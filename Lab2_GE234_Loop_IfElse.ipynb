## 🔹 ตัวอย่างที่ 1: การใช้ `for loop` เพื่อแสดงค่าพิกัด


```python

# พิกัดของสถานที่สำคัญในประเทศไทย
locations = {
    "Bangkok": (13.7563, 100.5018),
    "Chiang Mai": (18.7883, 98.9853),
    "Phuket": (7.8804, 98.3923),
    "Ubon Ratchathani": (15.2287, 104.8564)
}

# วนลูปแสดงค่าพิกัด
for city, coord in locations.items():
    print(f"{city}: Latitude {coord[0]}, Longitude {coord[1]}")
```


## 🔹 ตัวอย่างที่ 2: การใช้ `if-else` กรองข้อมูลพิกัด


```python
# สร้างข้อมูลจำลอง (ละติจูด, ลองจิจูด)
locations = {
    "เชียงใหม่": (18.7883, 98.9853),
    "เชียงราย": (19.9105, 99.8406),
    "กรุงเทพฯ": (13.7563, 100.5018),
    "ภูเก็ต": (7.8804, 98.3923),
    "แม่ฮ่องสอน": (19.3000, 97.9667)
}
for city, coord in locations.items():
    if coord[0] > 16:
        print(f"{city} อยู่ทางภาคเหนือของประเทศไทย")
    else:
        print(f"{city} ไม่ใช่จังหวัดทางภาคเหนือ")
```

locations เพื่อเก็บชื่อจังหวัดและค่าพิกัด (ละติจูด, ลองจิจูด)

ในภาษา Python ค่าพิกัดมักจะเก็บเป็น tuple หรือ list โดยในที่นี้โค้ดที่เช็คคือ coord[0] (ตัวเลขตัวแรก) ซึ่งหมายถึง Latitude (ละติจูด)


## 🔹 ตัวอย่างที่ 3: ใช้ `while loop` นับจำนวนจุดข้อมูล


```python

points = [(13.7563, 100.5018), (18.7883, 98.9853), (7.8804, 98.3923)]
count = 0

while count < len(points):
    print(f"Point {count + 1}: {points[count]}")
    count += 1
```


## 🔹 ตัวอย่างที่ 4: วิเคราะห์ข้อมูล Raster ด้วย `for loop`


```python

import numpy as np

# จำลองข้อมูล NDVI (Normalized Difference Vegetation Index)
ndvi = np.array([
    [0.2, 0.4, 0.5],
    [0.1, 0.3, 0.7],
    [0.6, 0.8, 0.2]
])

# วนลูปตรวจสอบค่าพืชพรรณ
for row in range(ndvi.shape[0]):
    for col in range(ndvi.shape[1]):
        if ndvi[row, col] > 0.3:
            print(f"พิกัด ({row}, {col}) มีพืชพรรณหนาแน่น (NDVI={ndvi[row, col]})")
```


# 📝 **กิจกรรมในแลป**

1. **แบบฝึกหัด 1**: เขียน `for loop` เพื่อแสดงชื่อจังหวัดและพิกัดจากลิสต์ของคุณเอง
2. **แบบฝึกหัด 2**: ใช้ `if-else` เพื่อแยกแยะพื้นที่ที่เป็นภูเขาและพื้นที่ราบจากลิสต์ของพิกัด
3. **แบบฝึกหัด 3**: ใช้ `while loop` เพื่อวนลูปและแสดงค่าละติจูดจากลิสต์จนกว่าจะพบค่าต่ำกว่า 10°
4. **แบบฝึกหัด 4**: ใช้ `for loop` วิเคราะห์ค่า NDVI และหาค่าพื้นที่ที่มีสภาพแห้งแล้ง (NDVI < 0.2)


```python
# 1: for loop เพื่อแสดงชื่อจังหวัดและพิกัดจากสิสต์

locations = {
    "Pathum Thani": (14.01972, 100.53500),
    "Nonthaburi": (13.5144, 100.5048),
    "Chonburi": (13.2143, 100.5845),
    "Nakhon Nayok": (14.2036, 101.2173)
}

# วนลูปแสดงค่าพิกัด
for city, coord in locations.items():
    print(f"{city}: Latitude {coord[0]}, Longitude {coord[1]}")
```

```python
# 2: if-else เพื่อแยกแยะพื้นที่ที่เป็นภูเขาและพื้นที่ราบจากลิสต์ของพิกัด

# ข้อมูลพิกัดในรูปแบบ { "ชื่อสถานที่": (lat, lon, elevation) }
locations = {
    "ดอยอินทนนท์": (18.58, 98.48, 2565),
    "ทุ่งกุลาร้องไห้": (15.42, 103.65, 130),
    "ภูกระดึง": (16.87, 101.83, 1316),
    "กรุงเทพฯ": (13.75, 100.50, 1.5),
    "เขาค้อ": (16.63, 101.00, 900)
}

for place, data in locations.items():
  elevation = data[2]

  if elevation > 500:
        print(f"⛰️ {place}: เป็นพื้นที่ภูเขา (สูง {elevation} m)")
  else:
        print(f"🌾 {place}: เป็นพื้นที่ราบ (สูง {elevation} m)")
```

```python
# 3: ใช้ while loop เพื่อวนลูปและแสดงค่าละติจูดจากลิสต์จนกว่าจะพบค่าต่ำกว่า 10°

degrees = [32, 25, 18, 15, 35, 12, 8, 20, 22, 15]
count = 0

while count < len(degrees):
  current_lat = degrees[count]

  if current_lat < 10:
        print(f"พบค่า {current_lat} ซึ่งต่ำกว่า 10° หยุดการทำงาน")
        break

  print(f"วันที่ {count}: {current_lat}°")

  count += 1
```

```python
# 4: for loop วิเคราะห์ค่า NDVI และหาค่าพื้นที่ที่มีสภาพแห้งแล้ง (NDVI < 0.2)

# รายชื่อพื้นที่และค่า NDVI ที่วัดได้
ndvi_data = {
    "Zone A": 0.45,
    "Zone B": 0.12,
    "Zone C": 0.68,
    "Zone D": 0.05,
    "Zone E": 0.19,
    "Zone F": 0.33
}

drought_areas = [] # ลิสต์สำหรับเก็บพื้นที่ที่แห้งแล้ง

print("--- ผลการวิเคราะห์สภาพพื้นที่ ---")

for zone, ndvi in ndvi_data.items():
    if ndvi < 0.2:
        status = "🏜️ แห้งแล้ง/ดินว่างเปล่า"
        drought_areas.append(zone)
    elif 0.2 <= ndvi <= 0.5:
        status = "🌿 พืชพรรณเบาบาง"
    else:
        status = "🌳 ป่าไม้หนาแน่น"

    print(f"{zone}: ค่า NDVI = {ndvi:.2f} -> {status}")

# สรุปผลตอนท้าย
print("---")
print(f"พบพื้นที่แห้งแล้งทั้งหมด {len(drought_areas)} แห่ง ได้แก่: {', '.join(drought_areas)}")
```

drought_areas: สร้าง List ว่าง เพื่อใช้เป็นตะกร้าเก็บรายชื่อโซนที่วิเคราะห์แล้วพบว่า "แห้งแล้ง"

for zone, ndvi in ndvi_data.items():
ใช้คำสั่ง .items() เพื่อดึงทั้ง ชื่อโซน (zone) และ ค่าตัวเลข (ndvi) ออกมาทีละคู่จนครบทุกโซนใน Dictionary

print(f"พบพื้นที่แห้งแล้งทั้งหมด {len(drought_areas)} แห่ง ได้แก่: {', '.join(drought_areas)}")
len(drought_areas): นับจำนวนสมาชิกในลิสต์ว่ามีกี่โซนที่แห้งแล้ง
