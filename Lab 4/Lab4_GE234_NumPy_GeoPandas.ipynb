## 🔹 ตัวอย่างที่ 1: ใช้ NumPy คำนวณข้อมูล NDVI จากภาพ Raster


```python

import numpy as np

# สร้างข้อมูลตัวอย่างสำหรับภาพ NDVI (Normalized Difference Vegetation Index)
nir = np.array([[0.7, 0.8, 0.6], [0.9, 0.5, 0.3], [0.4, 0.7, 0.2]])
red = np.array([[0.3, 0.4, 0.2], [0.5, 0.3, 0.1], [0.2, 0.3, 0.1]])

# คำนวณค่า NDVI
ndvi = (nir - red) / (nir + red)

print("ค่า NDVI:")
print(ndvi)
```


## 🔹 ตัวอย่างที่ 2: ใช้ NumPy คำนวณค่าเฉลี่ยและค่ามากสุดของ NDVI


```python

print(f"ค่าเฉลี่ย NDVI: {np.mean(ndvi):.2f}")
print(f"ค่า NDVI สูงสุด: {np.max(ndvi):.2f}")
print(f"ค่า NDVI ต่ำสุด: {np.min(ndvi):.2f}")
```


# 📝 **กิจกรรมในแลป**

1. **แบบฝึกหัด 1**: ใช้ NumPy คำนวณค่า **Mean, Max, Min** ของค่า NDVI ในอาร์เรย์ที่สร้างขึ้นเอง
2. **แบบฝึกหัด 2**: ใช้ GeoPandas โหลด **Shapefile** ของจังหวัด และคำนวณพื้นที่ของแต่ละจังหวัด
3. **แบบฝึกหัด 3**: ใช้ GeoPandas ทำ **Spatial Join** ระหว่างข้อมูลจังหวัดและอำเภอ
4. **แบบฝึกหัด 4**: ใช้ NumPy และ GeoPandas ร่วมกันเพื่อหาข้อมูลจังหวัดที่มี NDVI เฉลี่ยสูงสุด


```python
#แบบฝึกหัด 1: ใช้ NumPy คำนวณค่า Mean, Max, Min ของค่า NDVI ในอาร์เรย์ที่สร้างขึ้นเอง

import numpy as np

# สร้างข้อมูลตัวอย่างสำหรับภาพ NDVI (Normalized Difference Vegetation Index)
nir = np.array([[0.7, 0.8, 0.6], [0.9, 0.5, 0.3], [0.4, 0.7, 0.2]])
red = np.array([[0.3, 0.4, 0.2], [0.5, 0.3, 0.1], [0.2, 0.3, 0.1]])

# คำนวณค่า NDVI
ndvi = (nir - red) / (nir + red)
print("NDVI Array:")
print(np.round(ndvi, 2))
print("_" * 40)

mean_val = np.mean(ndvi)
max_val = np.max(ndvi)
min_val = np.min(ndvi)

print(f"\nMean (ค่าเฉลี่ย): {mean_val:.4f}") #\n เว้น 1 บรรทัด
print(f"Max (ค่าสูงสุด): {max_val:.4f}")
print(f"Min (ค่าต่ำสุด): {min_val:.4f}")
```

```python
#แบบฝึกหัด 2: ใช้ GeoPandas โหลด Shapefile ของจังหวัด และคำนวณพื้นที่ของแต่ละจังหวัด

import geopandas as gpd
import pandas as pd

# อ่านไฟล์ Shapefile และเก็บไว้ใน gdf เพื่อให้สอดคล้องกับการใช้งานต่อไปในเซลล์
shapefile_path = '/content/drive/MyDrive/GE234/Amphoe'
gdf = gpd.read_file(shapefile_path)

# เปลี่ยนชื่อคอลัมน์จาก P_NAME_T เป็น AMPHOE
gdf = gdf.rename(columns={'P_NAME_T': 'AMPHOE'})

# ตรวจสอบค่า CRS (Coordinate Reference System)
print(gdf.crs)

# คำนวณพื้นที่ของแต่ละจังหวัด (หน่วยเป็นตารางเมตร)
gdf["area_sqkm"] = gdf.geometry.area / 1e6
top_5 = gdf.nlargest(5, "area_sqkm")[["AMPHOE", "area_sqkm"]]

# รีเซ็ต Index ใหม่
top_5 = top_5.reset_index(drop=True)
top_5.index = top_5.index + 1

# แสดงทศนิยม 4 ตำแหน่ง
pd.options.display.float_format = '{:.4f}'.format

# แสดงพื้นที่จังหวัด 5 อันดับแรกที่ใหญ่ที่สุด
print("ผลการคำนวณพื้นที่จังหวัด (5 อันดับแรก):")
print(top_5)
```

```python
# แบบฝึกหัด 3: ใช้ GeoPandas ทำ Spatial Join ระหว่างข้อมูลจังหวัดและอำเภอ

# โหลดข้อมูลชั้นข้อมูลอำเภอ
gdf_districts = gpd.read_file("/content/drive/MyDrive/GE234/Amphoe")

# ทำ Spatial Join ระหว่างอำเภอกับจังหวัด
gdf_joined = gpd.sjoin(gdf_districts, gdf, how="inner", predicate="within")

# ลบคอลัมน์ที่ไม่จำเป็นออก (Column Cleaning)
# ลบ index_right และคอลัมน์ที่ซ้ำซ้อนฝั่ง right ออก
cols_to_drop = ['index_right', 'P_NAME_E_right', 'A_NAME_T_right', 'A_NAME_E_right', 'Area_km2_right']
gdf_joined = gdf_joined.drop(columns=cols_to_drop)

# 3. เปลี่ยนชื่อคอลัมน์ (ลบคำว่า _left ออก)
gdf_joined = gdf_joined.rename(columns={
    'P_NAME_E_left': 'P_NAME_E',
    'A_NAME_T_left': 'A_NAME_T',
    'A_NAME_E_left': 'A_NAME_E',
    'Area_km2_left': 'Area_km2'
})
# แสดงตัวอย่างข้อมูลที่เชื่อมโยงกัน
print(gdf_joined.head())
```

```python
# แบบฝึกหัด 4: ใช้ NumPy และ GeoPandas ร่วมกันเพื่อหาข้อมูลจังหวัดที่มี NDVI เฉลี่ยสูงสุด
```

```python
import pandas as pd
import numpy as np
import geopandas as gpd

# โหลดข้อมูล NDVI ของจังหวัด (ไฟล์ CSV)
csv_file = "/content/drive/MyDrive/GE234/tha-ndvi-subnat-full.csv"
df = pd.read_csv(csv_file)

# เลือกเฉพาะข้อมูลระดับจังหวัด (adm_level = 1)
df_prov = df[df['adm_level'] == 1].copy()

# คำนวณค่าเฉลี่ย NDVI ของแต่ละจังหวัด
mean_ndvi = df_prov.groupby('PCODE')['vim_avg'].mean().reset_index()

# สร้างรหัสจังหวัดให้ตรงกับ shapefile (ตัด "TH" ออก)
mean_ndvi['pro_code'] = mean_ndvi['PCODE'].str.replace('TH', '')

# โหลดข้อมูลขอบเขตจังหวัด (GeoJSON)
geojson_file = "/content/drive/MyDrive/GE234/provinces.geojson"
gdf_map = gpd.read_file(geojson_file)

# ทำ Attribute Join
gdf_join = gdf_map.merge(mean_ndvi, left_on='pro_code', right_on='pro_code', how='left')

# หาจังหวัดที่ NDVI เฉลี่ยสูงสุด
top_ndvi = gdf_join.nlargest(1, 'vim_avg')

print(" ___ รายงานข้อมูล NDVI ❤ ___ ")
print(f"\nจังหวัดที่มีค่า NDVI เฉลี่ยสูงสุด: {top_ndvi.iloc[0]['pro_th']}")
print(f"รหัสจังหวัด: {top_ndvi.iloc[0]['PCODE']}")
print(f"ค่า NDVI เฉลี่ย: {top_ndvi.iloc[0]['vim_avg']:.4f}")

# แสดง 5 อันดับจังหวัดที่ NDVI สูงสุด
top_5 = gdf_join.nlargest(5, 'vim_avg')[['pro_th', 'PCODE', 'vim_avg']]
top_5 = top_5.reset_index(drop=True)
top_5.index = top_5.index + 1

# เปลี่ยนชื่อคอลัมน์ (Rename)
top_5 = top_5.rename(columns={
    'pro_th': 'จังหวัด',
    'PCODE': 'รหัสจังหวัด',
    'vim_avg': 'ค่า NDVI เฉลี่ย'
})

print("_" * 40)
print("\n 5 อันดับจังหวัดที่มีค่า NDVI สูงสุด ")
print(top_5)
```
